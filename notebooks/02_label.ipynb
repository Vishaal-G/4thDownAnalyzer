```python
import pandas as pd
df = pd.read_parquet("../data/raw_4th_downs.parquet")

df = df[(df['play_type'] != 'no_play') & (df['play_type'] != 'qb_kneel') &  df['play_type'].notna()]
df['play_type'].value_counts(dropna=False)

#Add new column to indictate GO, FIELD_GOAL or PUNT
df['decision'] = df['play_type'].map({'punt': 'PUNT', 'pass': 'GO', 'run': 'GO', 'field_goal': 'FIELD_GOAL'})
df['decision'].value_counts()
```

```python
#Trim columns to what I need and get rid of anything that doesnt incloude epa
clean_fourth_downs = df[['yardline_100', 'ydstogo', 'score_differential', 'game_seconds_remaining', 'decision', 'epa', 'season']]
clean_fourth_downs.isna().sum()

clean_fourth_downs.to_parquet("../data/clean_4th_downs.parquet")
```
