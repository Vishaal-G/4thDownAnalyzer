```python
import nfl_data_py as nfl
pbp = nfl.import_pbp_data(years=range(2014,2024), downcast=True)
```

```python
pbp.shape
pbp.columns
pbp.dtypes.value_counts()
```

```python
fourth_downs = pbp[pbp['down'] == 4]
fourth_downs['play_type'].value_counts()

fourth_downs[['yardline_100', 'ydstogo', 'score_differential', 'game_seconds_remaining']].describe()

maxYDs = fourth_downs[fourth_downs['ydstogo'] == fourth_downs['ydstogo'].max()]
print(list(maxYDs['desc']))
```

```python
fourth_downs['play_type'].isna().sum()

#Store the fourth down plays as a parquet (preserves column by column data types)
fourth_downs.to_parquet("../data/raw_4th_downs.parquet")
```
